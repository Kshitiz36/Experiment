# Session to Create a linked list

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
  
node1= Node(10)
node2= Node(20)
node3= Node(30)
node4= Node(40)

head=node1

node1.next=node2
node2.next=node3
node3.next=node4


print(head.data)
print(head.next.data)
print(head.next.next.data)


# printing the linked list
def PRINT_LINKED_LIST(head):
    curr=head
    while curr!=None:
        print(curr.data , end='--->')
        curr = curr.next

```

```python
PRINT_LINKED_LIST(head)
```
